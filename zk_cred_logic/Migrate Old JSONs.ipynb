```python
import os
import json

# Define the directory
directory = "test"

# List to store parsed JSON data
json_list = []

# Loop through files in the directory
for filename in os.listdir(directory):
    if filename.endswith(".json"):  # Ensure it's a JSON file
        filepath = os.path.join(directory, filename)
        with open(filepath, "r", encoding="utf-8") as file:
            data = json.load(file)  # Parse JSON into a dictionary
            json_list.append(data)

# Now, json_list contains all the parsed JSON dictionaries
test_json = json_list[0]
```

```python
def ascii_list_to_str(l):
    return ''.join(chr(i) for i in l)
```

```python
def ip_to_field_element(ip):
    if ip == "":
        return 0
    parts = ip.split('.')
    assert(len(parts) == 4 and all(0 <= int(part) <= 255 for part in parts))
    felt = 0
    pow_of_256 = 1
    parts.reverse()
    for part in parts:
        felt += int(part) * pow_of_256
        pow_of_256 *= 256
    return felt

def hal_ip_format_to_field_element(lst):
    if isinstance(lst, int):
        return lst
    ip_str = ascii_list_to_str(lst).lstrip('0')
    return ip_to_field_element(ip_str)
```

```python
def reprocess_json(json_elem):
    json_elem['nonce'] = int(json_elem['nonce'])
    json_elem['is_challenge_response'] = int(json_elem['is_challenge_response'])
    json_elem['new_ip'] = hal_ip_format_to_field_element(json_elem['new_ip'])
    for i in range(0, len(json_elem['ipStrings'])):
        json_elem['ipStrings'][i] = hal_ip_format_to_field_element(json_elem['ipStrings'][i])
    
```

```python
for filename in os.listdir(directory):
        if filename.endswith(".json"):  # Ensure it's a JSON file
            filepath = os.path.join(directory, filename)
            
            # Read the JSON file
            with open(filepath, "r", encoding="utf-8") as file:
                data = json.load(file)  # Parse JSON into a dictionary
            
            # Apply the reprocess_json function
            reprocess_json(data)
            
            # Write back the modified JSON
            with open(filepath, "w", encoding="utf-8") as file:
                json.dump(data, file, indent=4)  # Pretty-print with indentation
```
